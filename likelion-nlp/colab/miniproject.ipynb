```python
import json

output_filename = "fine_tuning_dataset.jsonl"
print("📅 Gemma-2 전용 규격(system 역할 제거)으로 파인튜닝 데이터셋을 재구성합니다...")

with open(output_filename, "w", encoding="utf-8") as f:
    for ch in SUB_CHAPTERS:
        member = ch["member"]
        profile = TEAM_PROFILES.get(member, {"mbti": "ALL", "motto": "다같이 파이팅!"})
        keywords = ch["keywords"]
        
        system_instruction = "[지시사항: 너는 스터디 목차와 담당자를 완벽하게 꿰고 있는 스터디 마스터 봇이야. 아래 질문을 받으면 매칭되는 단원 정보와 담당 팀원 프로필을 말해주고 관련 퀴즈를 출제해야 해.]\n\n"
        user_msg = system_instruction + f"{ch['section']} {ch['title']} 파트 중에서 '{keywords}' 개념이 잘 이해가 안 가요. 설명해주세요!"
        
        assistant_msg = f"안녕하세요! 이 파트는 [소단원 {ch['section']}: {ch['title']}]에 해당하며, 담당 스터디원은 {member}님(MBTI: {profile['mbti']})입니다. {member}님의 한마디는 '{profile['motto']}'입니다! 자, 공부하신 {keywords}와 관련된 퀴즈입니다: '이 개념의 핵심 특징은 무엇일까요?' 정답과 해설은 교재를 참고하세요!"
        
        jsonl_line = {
            "messages": [
                {"role": "user", "content": user_msg},
                {"role": "assistant", "content": assistant_msg}
            ]
        }
        f.write(json.dumps(jsonl_line, ensure_ascii=False) + "\n")

print(f"✅ Gemma-2 맞춤형 '{output_filename}' 파일 생성이 완료되었습니다!")
```

```python
from huggingface_hub import notebook_login
notebook_login()

import os
import torch
from datasets import load_dataset
from transformers import AutoModelForCausalLM, AutoTokenizer, BitsAndBytesConfig
from trl import SFTTrainer, SFTConfig 
from peft import LoraConfig

model_id = "google/gemma-2-2b-it" 


current_working_dir = os.getcwd() 
local_output_dir = os.path.join(current_working_dir, "saved_study_bot")

print("📦 구글 서버로부터 Gemma-2-2b-it 모델 로드 중...")
tokenizer = AutoTokenizer.from_pretrained(model_id)
tokenizer.pad_token = tokenizer.eos_token

bnb_config = BitsAndBytesConfig(
    load_in_4bit=True,
    bnb_4bit_quant_type="nf4",
    bnb_4bit_compute_dtype=torch.bfloat16 
)

model = AutoModelForCausalLM.from_pretrained(
    model_id,
    quantization_config=bnb_config,
    device_map="auto"
)

peft_config = LoraConfig(
    r=8,
    lora_alpha=16,
    target_modules=["q_proj", "v_proj", "k_proj", "o_proj"], 
    lora_dropout=0.05,
    bias="none",
    task_type="CAUSAL_LM"
)

dataset = load_dataset("json", data_files="fine_tuning_dataset.jsonl", split="train")

sft_config = SFTConfig(
    output_dir=local_output_dir,        
    per_device_train_batch_size=4,       
    gradient_accumulation_steps=2,
    learning_rate=2e-4,                  
    logging_steps=5,                     
    num_train_epochs=3,                  
    bf16=True,                           
    save_strategy="no",                  
    report_to="none"
)

print("🔥Gemma 모델 파인튜닝 학습을 시작합니다...")
trainer = SFTTrainer(
    model=model,
    train_dataset=dataset,
    peft_config=peft_config,
    args=sft_config, 
)

# 1. 파인튜닝 학습 진행
trainer.train()

print(f"\n💾 내 로컬 컴퓨터 탐색기 경로로 가중치 직접 저장 중...\n👉 {local_output_dir}")
trainer.model.save_pretrained(local_output_dir)

print("\n🎉 모든 작업이 드디어 성공적으로 완수되었습니다!")
print("💡 VS Code 탐색기에서 'colab' 폴더 옆 새로고침 버튼을 눌러보세요!")
```
